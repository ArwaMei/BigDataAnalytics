```python
# Example input text
text = """
spark makes big data easy
big data needs spark power
python is great for data processing
"""
```

```python
def mapper(line):
    words = line.lower().split()
    return [(word, 1) for word in words]
```

```python
mapped = []

for line in text.strip().split("\n"):
    mapped.extend(mapper(line))

print(mapped[:20])  # preview first outputs
```

```python
from collections import defaultdict

grouped = defaultdict(list)

for word, count in mapped:
    grouped[word].append(count)

# Show grouped structure
for key in list(grouped.keys())[:10]:
    print(key, grouped[key])
```

```python
def reducer(key, values):
    return (key, sum(values))
```

```python
reduced = []

for key, values in grouped.items():
    reduced.append(reducer(key, values))

# Sort output alphabetically
reduced = sorted(reduced)

for word, count in reduced:
    print(word, ":", count)
```
