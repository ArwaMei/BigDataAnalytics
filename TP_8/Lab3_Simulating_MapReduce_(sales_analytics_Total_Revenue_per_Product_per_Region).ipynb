```python
def mapper(line):
    fields = line.strip().split(",")
    if len(fields) != 4:
        return []  # skip invalid lines

    product, qty, price, region = fields
    revenue = int(qty) * float(price)

    return [(region, revenue)]
```

```python
mapped = []

for line in data.strip().split("\n"):
    mapped.extend(mapper(line))

mapped
```

```python
from collections import defaultdict

grouped = defaultdict(list)

for region, value in mapped:
    grouped[region].append(value)

grouped
```

```python
def reducer(region, values):
    return (region, sum(values))
```

```python
reduced = []

for region, values in grouped.items():
    reduced.append(reducer(region, values))

# Sort alphabetically for nicer display
reduced = sorted(reduced)

for region, total in reduced:
    print(f"{region}: {total}")
```
