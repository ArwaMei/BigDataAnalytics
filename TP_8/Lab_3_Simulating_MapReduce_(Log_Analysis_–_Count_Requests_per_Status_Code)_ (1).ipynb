Implement the Mapper

```python
# Mapper: Extract (StatusCode, 1)
def mapper(line):
    fields = line.strip().split(",")
    if len(fields) != 7 or fields[0].startswith('#'):
        return []
    # TODO: extract the status code (5th or 6th field)
    status = fields[5]
    return [(status, 1)]
```

Shuffle Phase

```python
from collections import defaultdict

def shuffle(mapped_data):
    grouped = defaultdict(list)
    for key, value in mapped_data:
        grouped[key].append(value)
    return grouped
```

Reducer Phase

```python
from collections import defaultdict

def reducer(mapped_data):
    grouped = defaultdict(int)
    for key, value in mapped_data:
        grouped[key] += value
    return grouped

```

Combine the Phases

```python
mapped = []
with open("weblogs.txt", "r") as f:
    for line in f:
        mapped.extend(mapper(line))

reduced = reducer(mapped)

for code, count in sorted(reduced.items()):
    print(f"HTTP {code}: {count} requests")
```

```python
def mapper_url(line):
    fields = line.strip().split(",")
    if len(fields) != 7 or fields[0].startswith("#"):
        return []
    url = fields[4].strip()
    return [(url, 1)]

mapped = []
with open("weblogs.txt", "r") as f:
    for line in f:
        mapped.extend(mapper_url(line))

reduced = reducer(mapped)

print("Requests per URL ")
for url, count in sorted(reduced.items()):
    print(f"{url}: {count} requests")
```

```python
def mapper_size(line):
    fields = line.strip().split(",")
    if len(fields) != 7 or fields[0].startswith("#"):
        return []

    status = fields[5].strip()
    size = int(fields[6].strip())

    return [(status, size)]

mapped = []
with open("weblogs.txt", "r") as f:
    for line in f:
        mapped.extend(mapper_size(line))

reduced = reducer(mapped)

print("Total Response Size per Status Code")
for code, total_size in sorted(reduced.items()):
    print(f"HTTP {code}: {total_size} bytes")
```

```python
def mapper_errors_only(line):
    fields = line.strip().split(",")
    if len(fields) != 7 or fields[0].startswith("#"):
        return []

    status = fields[5].strip()

    if status == "200":
        return []

    return [(status, 1)]

mapped = []
with open("weblogs.txt", "r") as f:
    for line in f:
        mapped.extend(mapper_errors_only(line))

reduced = reducer(mapped)

print("Error requests")
for code, count in sorted(reduced.items()):
    print(f"HTTP {code}: {count} errors")
```
