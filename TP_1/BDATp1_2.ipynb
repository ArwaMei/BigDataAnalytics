```python
# -----------------------------------------------
# 📚 Web Scraping Project Gutenberg Books
# -----------------------------------------------

import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

base_url = "https://www.gutenberg.org/ebooks/search/?sort_order=downloads&start_index={}"

titles = []
authors = []
downloads = []

for i in range(1, 1001, 25):
    print(f"Scraping books {i}–{i+24}...")
    url = base_url.format(i)

    response = requests.get(url)
    soup = BeautifulSoup(response.text, "html.parser")

    books = soup.find_all("li", class_="booklink")

    for book in books:
        title = book.find("span", class_="title").get_text(strip=True)
        author_tag = book.find("span", class_="subtitle")
        author = author_tag.get_text(strip=True) if author_tag else "Unknown"

        extra = book.find("span", class_="extra")
        download_count = extra.get_text(strip=True).replace("downloads", "").strip() if extra else "N/A"

        titles.append(title)
        authors.append(author)
        downloads.append(download_count)

    time.sleep(1)  

df = pd.DataFrame({
    "Title": titles,
    "Author": authors,
    "Downloads": downloads
})

df.to_csv("gutenberg_books.csv", index=False, encoding="utf-8")

print("\n✅ Scraping completed successfully!")
print(f"Total books scraped: {len(df)}")
df.head()
```
